```python
import random
import hashlib

# Configuration
SURNAME = "Espinase"
SEED_NUM = 7
NUM_READINGS = 10

# Operating ranges for classification
RANGES = {
    'Low': (0, 30),
    'Normal': (30, 70),
    'High': (70, 90),
    'Critical': (90, 100)
}

# Generate sensor data using surname and seed
def generate_sensor_data(surname, seed_num, count):
    """Generate unique sensor readings using surname and SEED_NUM"""
    random.seed(hash(surname) + seed_num)
    readings = []
    
    for i in range(count):
        # 80% valid readings, 20% invalid data
        if random.random() < 0.8:
            reading = round(random.uniform(0, 100), 2)
            readings.append(reading)
        else:
            # Invalid data (string or out of range)
            readings.append("ERROR")
    
    return readings

# Validate a reading
def validate_reading(reading):
    """Validate if reading is a valid number"""
    try:
        value = float(reading)
        if 0 <= value <= 100:
            return True, value
        else:
            return False, None
    except (ValueError, TypeError):
        return False, None

# Classify a valid reading
def classify_reading(value):
    """Classify reading based on operating ranges"""
    for category, (min_val, max_val) in RANGES.items():
        if min_val <= value < max_val:
            return category
    return 'Critical'

# Main program
print("=" * 60)
print("SENSOR READING VALIDATION & CLASSIFICATION SYSTEM")
print("=" * 60)
print(f"Surname: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Total Readings to Process: {NUM_READINGS}")
print("=" * 60)

# Generate sensor data
raw_readings = generate_sensor_data(SURNAME, SEED_NUM, NUM_READINGS)

print("\nGenerated Sensor Data:")
print(raw_readings)
print()

# Process readings
valid_count = 0
invalid_count = 0
classifications = {'Low': 0, 'Normal': 0, 'High': 0, 'Critical': 0}
valid_readings_list = []

print("-" * 60)
print("VALIDATION & CLASSIFICATION RESULTS")
print("-" * 60)

for i, reading in enumerate(raw_readings, 1):
    is_valid, value = validate_reading(reading)
    
    if is_valid:
        classification = classify_reading(value)
        valid_count += 1
        classifications[classification] += 1
        valid_readings_list.append((value, classification))
        status = "✓ VALID"
        print(f"Reading {i:2d}: {value:6.2f} → {classification:8s} {status}")
    else:
        invalid_count += 1
        status = "✗ INVALID"
        print(f"Reading {i:2d}: {str(reading):6s} → {'N/A':8s} {status}")

# Summary Report
print("\n" + "=" * 60)
print("SUMMARY REPORT")
print("=" * 60)
print(f"Total Readings Processed: {NUM_READINGS}")
print(f"Valid Readings:           {valid_count}")
print(f"Invalid Readings:         {invalid_count}")
print(f"Success Rate:             {(valid_count/NUM_READINGS)*100:.1f}%")

print("\nClassification Breakdown:")
for category in ['Low', 'Normal', 'High', 'Critical']:
    count = classifications[category]
    percentage = (count / valid_count * 100) if valid_count > 0 else 0
    print(f"  {category:8s}: {count:2d} readings ({percentage:5.1f}%)")

print("=" * 60)
```
