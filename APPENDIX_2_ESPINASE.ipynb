```python
import random
import string

# Configuration
SURNAME = "Espinase"
SEED_NUM = 7

# Generate signal using surname and SEED_NUM
def generate_signal(surname, seed_num):
    """Generate a unique signal using surname and SEED_NUM"""
    random.seed(hash(surname) + seed_num)
    
    # Create a pool of characters
    char_pool = string.ascii_letters + string.digits + " !@#$%^&*()"
    
    # Generate signal of varying length
    signal_length = len(surname) * seed_num + random.randint(5, 15)
    signal = ''.join(random.choice(char_pool) for _ in range(signal_length))
    
    return signal

# Process and normalize signal
def process_signal(signal):
    """Process and normalize the signal"""
    # Remove leading/trailing whitespace
    processed = signal.strip()
    # Normalize spaces (replace multiple spaces with single space)
    processed = ' '.join(processed.split())
    return processed

# Analyze signal character by character
def analyze_signal(signal):
    """Analyze signal character by character"""
    analysis = {
        'total_characters': len(signal),
        'letters': {'count': 0, 'uppercase': 0, 'lowercase': 0, 'chars': []},
        'digits': {'count': 0, 'chars': []},
        'spaces': {'count': 0},
        'special': {'count': 0, 'chars': []},
    }
    
    char_details = []
    
    for i, char in enumerate(signal, 1):
        if char.isalpha():
            analysis['letters']['count'] += 1
            if char.isupper():
                analysis['letters']['uppercase'] += 1
            else:
                analysis['letters']['lowercase'] += 1
            analysis['letters']['chars'].append(char)
            char_type = "LETTER"
            subtype = "Uppercase" if char.isupper() else "Lowercase"
        elif char.isdigit():
            analysis['digits']['count'] += 1
            analysis['digits']['chars'].append(char)
            char_type = "DIGIT"
            subtype = char
        elif char == ' ':
            analysis['spaces']['count'] += 1
            char_type = "SPACE"
            subtype = "Whitespace"
        else:
            analysis['special']['count'] += 1
            analysis['special']['chars'].append(char)
            char_type = "SPECIAL"
            subtype = char
        
        char_details.append({
            'position': i,
            'character': char,
            'type': char_type,
            'subtype': subtype,
            'ascii': ord(char)
        })
    
    return analysis, char_details

# Classify signal
def classify_signal(analysis):
    """Classify signal based on characteristics"""
    total = analysis['total_characters']
    letter_ratio = analysis['letters']['count'] / total if total > 0 else 0
    digit_ratio = analysis['digits']['count'] / total if total > 0 else 0
    special_ratio = analysis['special']['count'] / total if total > 0 else 0
    
    if letter_ratio > 0.6:
        classification = "TEXT SIGNAL"
        confidence = "HIGH"
    elif digit_ratio > 0.6:
        classification = "NUMERIC SIGNAL"
        confidence = "HIGH"
    elif letter_ratio > 0.3 and digit_ratio > 0.3:
        classification = "ALPHANUMERIC SIGNAL"
        confidence = "HIGH"
    elif special_ratio > 0.4:
        classification = "ENCODED/SPECIAL SIGNAL"
        confidence = "MEDIUM"
    else:
        classification = "MIXED SIGNAL"
        confidence = "MEDIUM"
    
    return classification, confidence

# Main Program
print("=" * 70)
print("SIGNAL ANALYSIS & DIAGNOSTIC SYSTEM")
print("=" * 70)
print(f"Surname: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
print("=" * 70)

# Generate signal
raw_signal = generate_signal(SURNAME, SEED_NUM)
print(f"\n📡 Generated Signal (Raw):")
print(f"   {raw_signal}")
print(f"   Length: {len(raw_signal)} characters")

# Process signal
processed_signal = process_signal(raw_signal)
print(f"\n⚙️  Processed Signal (Normalized):")
print(f"   {processed_signal}")
print(f"   Length: {len(processed_signal)} characters")

# Analyze signal
analysis, char_details = analyze_signal(processed_signal)

print(f"\n📊 Character Analysis:")
print(f"   Total Characters: {analysis['total_characters']}")
print(f"   Letters: {analysis['letters']['count']} (U: {analysis['letters']['uppercase']}, L: {analysis['letters']['lowercase']})")
print(f"   Digits: {analysis['digits']['count']}")
print(f"   Spaces: {analysis['spaces']['count']}")
print(f"   Special Characters: {analysis['special']['count']}")

# Detailed character-by-character breakdown
print(f"\n🔍 Character-by-Character Breakdown:")
print(f"{'Pos':>4} {'Char':>6} {'Type':>12} {'Subtype':>15} {'ASCII':>6}")
print("-" * 50)
for detail in char_details[:20]:  # Show first 20 characters
    char_display = detail['character'] if detail['character'] != ' ' else '␣'
    print(f"{detail['position']:4d} {char_display:>6} {detail['type']:>12} {detail['subtype']:>15} {detail['ascii']:>6}")

if len(char_details) > 20:
    print(f"... and {len(char_details) - 20} more characters")

# Classify signal
classification, confidence = classify_signal(analysis)

print(f"\n🎯 Signal Classification:")
print(f"   Classification: {classification}")
print(f"   Confidence: {confidence}")

print("\n" + "=" * 70)
print("DIAGNOSTIC REPORT COMPLETE")
print("=" * 70)
```
