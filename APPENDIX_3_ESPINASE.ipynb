```python
import random
import string
import hashlib

# Configuration
SURNAME = "Espinase"
SEED_NUM = 7

# Generate password using surname and SEED_NUM
def generate_password(surname, seed_num):
    """Generate a unique access credential using surname and SEED_NUM"""
    random.seed(hash(surname) + seed_num)
    
    # Create password from surname + seed variations
    password_length = len(surname) + seed_num
    
    # Mix of uppercase, lowercase, digits, and special characters
    char_pool = string.ascii_uppercase + string.ascii_lowercase + string.digits + "!@#$"
    password = ''.join(random.choice(char_pool) for _ in range(password_length))
    
    return password

# Determine attempt limit based on surname and seed
def determine_attempt_limit(surname, seed_num):
    """Determine the allowed attempt limit"""
    # Base limit: length of surname
    base_limit = len(surname)
    # Adjust based on seed
    attempt_limit = max(3, base_limit - (seed_num % 3))
    return attempt_limit

# Validate password attempt
def validate_attempt(attempt, correct_password):
    """Check if attempt matches the correct password"""
    return attempt.strip() == correct_password

# Simulate authentication process
def authenticate_system(correct_password, attempt_limit):
    """Simulate the authentication process with attempt management"""
    attempts_made = 0
    access_granted = False
    attempt_history = []
    
    # Simulate user attempts (for demonstration)
    # In a real system, this would be user input
    attempts_to_try = [
        "wrongpass1",
        "incorrect",
        correct_password,  # Correct password on 3rd attempt
    ]
    
    print(f"\n🔐 AUTHENTICATION PROCESS INITIATED")
    print(f"   Attempt Limit: {attempt_limit}")
    print("-" * 60)
    
    for attempt_num in range(1, attempt_limit + 1):
        attempts_made = attempt_num
        
        # Simulate getting an attempt
        if attempt_num <= len(attempts_to_try):
            user_attempt = attempts_to_try[attempt_num - 1]
        else:
            user_attempt = f"wrong{attempt_num}"
        
        # Validate attempt
        is_correct = validate_attempt(user_attempt, correct_password)
        attempt_history.append({
            'attempt_number': attempt_num,
            'attempt': user_attempt,
            'result': 'ACCEPTED' if is_correct else 'REJECTED',
            'is_correct': is_correct
        })
        
        if is_correct:
            access_granted = True
            print(f"   Attempt {attempt_num}/{attempt_limit}: ✓ CORRECT PASSWORD")
            print(f"   🔓 ACCESS GRANTED")
            break
        else:
            remaining = attempt_limit - attempt_num
            print(f"   Attempt {attempt_num}/{attempt_limit}: ✗ INCORRECT")
            print(f"      Remaining attempts: {remaining}")
            
            if remaining == 0:
                print(f"   ⚠️  ATTEMPT LIMIT REACHED - SYSTEM LOCKED")
    
    return access_granted, attempts_made, attempt_history

# Determine final system state
def determine_system_state(access_granted, attempts_made, attempt_limit):
    """Determine the final state of the system"""
    if access_granted:
        state = "UNLOCKED"
        status = "✓ AUTHORIZED"
        description = "System access granted to authenticated user"
    else:
        state = "LOCKED"
        status = "✗ DENIED"
        description = "System locked due to failed authentication attempts"
    
    return state, status, description

# Main Program
print("=" * 70)
print("ELECTRONIC AUTHENTICATION SYSTEM - SECURITY MODULE")
print("=" * 70)
print(f"System ID: {SURNAME.upper()}_SEC")
print(f"Configuration SEED: {SEED_NUM}")
print("=" * 70)

# Generate credentials
correct_password = generate_password(SURNAME, SEED_NUM)
attempt_limit = determine_attempt_limit(SURNAME, SEED_NUM)

print(f"\n🔑 CREDENTIALS GENERATED:")
print(f"   Generated Password: {correct_password}")
print(f"   Password Length: {len(correct_password)} characters")
print(f"   Attempt Limit: {attempt_limit} attempts allowed")

# Run authentication
access_granted, attempts_made, attempt_history = authenticate_system(correct_password, attempt_limit)

# Determine final system state
system_state, access_status, state_description = determine_system_state(
    access_granted, attempts_made, attempt_limit
)

# Detailed attempt history
print(f"\n📋 ATTEMPT HISTORY:")
print(f"{'Attempt':>8} {'Status':>12} {'Details':>40}")
print("-" * 60)
for history in attempt_history:
    attempt_num = history['attempt_number']
    result = history['result']
    attempt = history['attempt']
    print(f"{attempt_num:>8} {result:>12} Input: {attempt[:30]}")

# Final Report
print(f"\n" + "=" * 70)
print(f"AUTHENTICATION REPORT")
print("=" * 70)
print(f"Access Result: {access_status}")
print(f"Attempts Made: {attempts_made}/{attempt_limit}")
print(f"Final System State: {system_state}")
print(f"Status Description: {state_description}")

if access_granted:
    print(f"\n✓ SYSTEM STATUS: OPERATIONAL")
    print(f"  User has full system access")
else:
    print(f"\n✗ SYSTEM STATUS: SECURITY LOCKDOWN")
    print(f"  All access denied - contact system administrator")

print("\n" + "=" * 70)
print("AUTHENTICATION PROCESS COMPLETE")
print("=" * 70)
```
